Statistical  measures : mean,median,mode,standard deviation and variance with python code(with and without function)

MEAN


```python
data = [10, 20, 30, 40, 50]
total = 0
for i in data:
  total = total + i
mean = total / len(data)
print("Mean ", mean)
```

```python
# with function
def mean(data):
    total = 0
    for i in data:
        total = total + i
    return total / len(data)
data = [10, 20, 30, 40, 50]
print("Mean ", mean(data))
```

```python
# median
def median(data):
    data.sort()
    n = len(data)
    if n % 2 == 0:
        return (data[n//2 - 1] + data[n//2]) / 2
    else:
        return data[n//2]
data = [10, 20, 30, 40, 50]
print("Median =", median(data))
```

MODE


```python
def mode(data):
    max_count = 0
    mode_value = 0
    for i in data:
        count = 0
        for j in data:
            if i == j:
                count = count + 1
        if count > max_count:
            max_count = count
            mode_value = i
    return mode_value
data = [1, 2, 2, 3, 4, 2, 5]
print("Mode =", mode(data))
```

VARIANCE

```python
data = [10,20,30,40,50]
mean = sum(data)/len(data)
total = 0
for i in data:
    total = total + (i - mean) ** 2
variance = total / len(data)
print("Variance =", variance)
```

```python
# VARIANCE WITH FUNCTION
def variance(data):
    mean = sum(data) / len(data)
    total = 0
    for i in data:
        total = total + (i - mean) ** 2
    return total / len(data)
data = [1,2,3,4,5]
print("Variance =", variance(data))
```
