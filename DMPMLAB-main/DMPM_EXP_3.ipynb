```python
# Data visualization is the process of representing data in the form of graphs, charts, and plots
# LINE PLOTS : A line plot is used to show trends or changes over a continuous or ordered variable, especially time.
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [1000, 1200, 1500, 1800, 2000]
plt.plot(months, sales)
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Sales Over Time")
plt.show()
```

```python
# A bar plot compares values between different categories.

import matplotlib.pyplot as plt

subjects = ["Math", "Science", "English", "Computer"]
marks = [85, 78, 92, 95]

plt.bar(subjects, marks)
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python

# A scatter plot displays the relationship between two numerical variables.
import matplotlib.pyplot as plt

height = [150, 155, 160, 165, 170]
weight = [50, 55, 58, 63, 68]

plt.scatter(height, weight)
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Height vs Weight")
plt.show()
```

```python
# A histogram shows the distribution of numerical data by dividing values into intervals called bins.
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 60, 62, 65, 68, 70, 75, 80, 85]

plt.hist(marks, bins=5)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Distribution of Marks")
plt.show()
```
